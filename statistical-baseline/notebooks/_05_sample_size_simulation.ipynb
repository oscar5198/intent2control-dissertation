# Phase 3C - Final-Design Sample-Size Simulation

This notebook replaces the old 3-mix / 5-context simulation with the final 5-mix, 3-episode, 20-stimulus design and the supervisor-aligned additive acoustic-feature model.

The scenarios are hypothetical design assumptions, not empirical listener results.

## Final Design and Model

- 4 songs, 2 groups, 2 non-overlapping songs per group
- 5 mixes per song, 20 unique stimuli
- 3 episodes: `EDR-1`, `EDR-2`, `FM-1`
- 30 ratings per participant
- Primary model: `rating ~ episode + group + z_RMS + z_CF + z_SW + (1 | participant_id) + (1 | stimulus_id)`

The notebook deliberately excludes episode interactions, random slopes, song/group random effects, `z_SI`, and Bark predictors from the primary simulation model.

```python

import os, json, math, time, hashlib, warnings
from pathlib import Path
from importlib import metadata

CACHE_ROOT = Path(os.environ.get("TEMP", ".")) / "phase3c_cache"
(CACHE_ROOT / "pytensor").mkdir(parents=True, exist_ok=True)
(CACHE_ROOT / "numba").mkdir(parents=True, exist_ok=True)
os.environ.setdefault("PYTENSOR_FLAGS", f"cxx=,base_compiledir={(CACHE_ROOT / 'pytensor').as_posix()}")
os.environ.setdefault("NUMBA_CACHE_DIR", str(CACHE_ROOT / "numba"))

import arviz as az
import bambi as bmb
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

warnings.filterwarnings("ignore", category=FutureWarning)
plt.style.use("seaborn-v0_8-whitegrid")

SEED = 55
def find_repo_root():
    start = Path.cwd().resolve()
    for candidate in [start, *start.parents]:
        feature_path = candidate / "statistical-baseline/outputs/feature_exploration/final_20_stimulus_feature_table.csv"
        if feature_path.exists():
            return candidate
    raise FileNotFoundError("Could not locate repository root containing the final Phase 3B.3 feature table.")
REPO_ROOT = find_repo_root()
FEATURE_TABLE_PATH = REPO_ROOT / "statistical-baseline/outputs/feature_exploration/final_20_stimulus_feature_table.csv"
OUTPUT_DIR = REPO_ROOT / "statistical-baseline/outputs/sample_size_simulation"
FIGURE_DIR = OUTPUT_DIR / "figures"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
FIGURE_DIR.mkdir(parents=True, exist_ok=True)

DESIGN_VERSION = "phase3c_final_5mix_3episode_20stimulus_additive_feature_v1"
EPISODES = ["EDR-1", "EDR-2", "FM-1"]
SAMPLE_SIZES = [20, 30, 40, 50, 60, 80]
REQUESTED_REPLICATIONS = 20
N_REPLICATIONS = 10
REPLICATION_FALLBACK_REASON = (
    "Fallback from 20 to 10 replications per condition because 360 repeated Bayesian fits "
    "was not computationally practical in this desktop execution."
)
FORMULA = "rating ~ episode + group + z_RMS + z_CF + z_SW + (1 | participant_id) + (1 | stimulus_id)"
SIM_CONFIG = dict(draws=400, tune=400, chains=2, cores=1, target_accept=0.95, inference_method="nutpie")
CONFIRM_CONFIG = dict(draws=1000, tune=1000, chains=4, cores=1, target_accept=0.95, inference_method="nutpie")

display(pd.DataFrame([
    {"component": "python", "version": os.sys.version.split()[0]},
    {"component": "pandas", "version": pd.__version__},
    {"component": "numpy", "version": np.__version__},
    {"component": "arviz", "version": az.__version__},
    {"component": "bambi", "version": bmb.__version__},
    {"component": "nutpie", "version": metadata.version("nutpie")},
]))
display(pd.DataFrame([{**SIM_CONFIG, "requested_replications": REQUESTED_REPLICATIONS, "actual_replications": N_REPLICATIONS}]))

feature_table = pd.read_csv(FEATURE_TABLE_PATH)
required = ["group", "song_id", "song_title", "stimulus_id", "mix_id", "original_mix_name", "z_RMS", "z_CF", "z_SW", "z_SI"]
missing = [c for c in required if c not in feature_table.columns]
if missing:
    raise ValueError(f"Missing feature columns: {missing}")
feature_table = feature_table.copy()
for c in ["group", "song_id", "stimulus_id", "mix_id"]:
    feature_table[c] = feature_table[c].astype(str)
checks = pd.DataFrame([
    {"check": "20 stimuli", "observed": feature_table.stimulus_id.nunique(), "expected": 20},
    {"check": "4 songs", "observed": feature_table.song_id.nunique(), "expected": 4},
    {"check": "2 groups", "observed": feature_table.group.nunique(), "expected": 2},
    {"check": "5 mixes per song", "observed": sorted(feature_table.groupby("song_id").stimulus_id.nunique().unique().tolist()), "expected": [5]},
    {"check": "complete z_RMS", "observed": bool(feature_table.z_RMS.notna().all()), "expected": True},
    {"check": "complete z_CF", "observed": bool(feature_table.z_CF.notna().all()), "expected": True},
    {"check": "complete z_SW", "observed": bool(feature_table.z_SW.notna().all()), "expected": True},
    {"check": "complete z_SI", "observed": bool(feature_table.z_SI.notna().all()), "expected": True},
])
checks["passed"] = checks.observed.astype(str).eq(checks.expected.astype(str))
if not checks.passed.all():
    raise AssertionError(checks)
display(checks)

BASE = dict(
    grand_mean=68.0,
    group_effects={"group_01": 0.0, "group_02": 1.2},
    episode_effects={"EDR-1": 0.0, "EDR-2": -1.4, "FM-1": 2.0},
    feature_effects={"z_RMS": 2.2, "z_CF": -1.7, "z_SW": 1.9},
    participant_sd=8.5,
    stimulus_sd=3.2,
    residual_sd=9.8,
)
SCENARIOS = {
    "moderate_expected": {**BASE, "description": "Moderate expected feature, episode, group, participant, stimulus and residual effects."},
    "smaller_effects_harder": {
        "description": "Fixed effects scaled to 60% of the moderate scenario; random and residual SDs unchanged.",
        "grand_mean": BASE["grand_mean"],
        "group_effects": {k: v * 0.6 for k, v in BASE["group_effects"].items()},
        "episode_effects": {k: v * 0.6 for k, v in BASE["episode_effects"].items()},
        "feature_effects": {k: v * 0.6 for k, v in BASE["feature_effects"].items()},
        "participant_sd": BASE["participant_sd"], "stimulus_sd": BASE["stimulus_sd"], "residual_sd": BASE["residual_sd"],
    },
    "higher_noise": {
        **{k: BASE[k] for k in ["grand_mean", "group_effects", "episode_effects", "feature_effects", "stimulus_sd"]},
        "description": "Moderate fixed effects with higher participant and residual variability.",
        "participant_sd": 11.0, "residual_sd": 12.5,
    },
}
FEATURE_TERMS = ["z_RMS", "z_CF", "z_SW"]
TRUE_TERMS = {
    "episode[EDR-2]": lambda s: s["episode_effects"]["EDR-2"] - s["episode_effects"]["EDR-1"],
    "episode[FM-1]": lambda s: s["episode_effects"]["FM-1"] - s["episode_effects"]["EDR-1"],
    "group[group_02]": lambda s: s["group_effects"]["group_02"] - s["group_effects"]["group_01"],
    "z_RMS": lambda s: s["feature_effects"]["z_RMS"],
    "z_CF": lambda s: s["feature_effects"]["z_CF"],
    "z_SW": lambda s: s["feature_effects"]["z_SW"],
}
def scenario_row(name, s):
    total = s["participant_sd"]**2 + s["stimulus_sd"]**2 + s["residual_sd"]**2
    return {
        "scenario": name, "description": s["description"],
        "group_02_minus_group_01": TRUE_TERMS["group[group_02]"](s),
        "episode_EDR_2_minus_EDR_1": TRUE_TERMS["episode[EDR-2]"](s),
        "episode_FM_1_minus_EDR_1": TRUE_TERMS["episode[FM-1]"](s),
        "z_RMS": s["feature_effects"]["z_RMS"], "z_CF": s["feature_effects"]["z_CF"], "z_SW": s["feature_effects"]["z_SW"],
        "participant_sd": s["participant_sd"], "stimulus_sd": s["stimulus_sd"], "residual_sd": s["residual_sd"],
        "participant_ICC": s["participant_sd"]**2 / total, "stimulus_ICC": s["stimulus_sd"]**2 / total, "residual_share": s["residual_sd"]**2 / total,
    }
scenario_table = pd.DataFrame([scenario_row(k, v) for k, v in SCENARIOS.items()])
display(scenario_table)

ADEQUACY_CRITERIA = [
    "operationally acceptable diagnostics: zero divergences, max R-hat <= 1.05, min bulk/tail ESS >= 50",
    "participant and stimulus ICC recovery are stable enough for variance-decomposition reporting",
    "primary feature-coefficient RMSE, coverage, and sign recovery are acceptable",
    "stimulus-by-episode expected-rating RMSE is low relative to the 0-100 scale",
    "preferred-mix winner accuracy and true-winner posterior probability are stable",
    "posterior uncertainty decreases with increasing N and larger samples show diminishing returns",
]
display(pd.DataFrame({"pre_specified_adequacy_criteria": ADEQUACY_CRITERIA}))

def participants(n):
    a = n // 2
    return pd.DataFrame(
        [{"participant_id": f"p{i+1:03d}", "group": "group_01"} for i in range(a)]
        + [{"participant_id": f"p{i+a+1:03d}", "group": "group_02"} for i in range(n-a)]
    )
def expected_table(s, stim_eff):
    rows = []
    for _, stim in feature_table.iterrows():
        for ep in EPISODES:
            feature = sum(s["feature_effects"][f] * stim[f] for f in FEATURE_TERMS)
            mu = s["grand_mean"] + s["group_effects"][stim.group] + s["episode_effects"][ep] + feature + stim_eff[stim.stimulus_id]
            rows.append({**stim[["song_id", "song_title", "group", "stimulus_id", "mix_id", "original_mix_name", "z_RMS", "z_CF", "z_SW"]].to_dict(), "episode": ep, "true_expected_rating": mu})
    return pd.DataFrame(rows)
def generate(n, scenario_name, rep):
    s = SCENARIOS[scenario_name]
    rg = np.random.default_rng(SEED + 100000 * (list(SCENARIOS).index(scenario_name) + 1) + 1000 * n + rep)
    ps = participants(n)
    p_eff = dict(zip(ps.participant_id, rg.normal(0, s["participant_sd"], len(ps))))
    st_eff = dict(zip(feature_table.stimulus_id, rg.normal(0, s["stimulus_sd"], len(feature_table))))
    exp = expected_table(s, st_eff)
    rows = []
    for _, p in ps.iterrows():
        for _, stim in feature_table[feature_table.group.eq(p.group)].iterrows():
            for ep in EPISODES:
                feature = sum(s["feature_effects"][f] * stim[f] for f in FEATURE_TERMS)
                pop_mu = s["grand_mean"] + s["group_effects"][p.group] + s["episode_effects"][ep] + feature + st_eff[stim.stimulus_id]
                raw = pop_mu + p_eff[p.participant_id] + rg.normal(0, s["residual_sd"])
                rows.append({**stim[["group", "song_id", "song_title", "stimulus_id", "mix_id", "original_mix_name", "z_RMS", "z_CF", "z_SW"]].to_dict(), "participant_id": p.participant_id, "episode": ep, "raw_rating": raw, "rating": float(np.clip(raw, 0, 100))})
    data = pd.DataFrame(rows)
    for c in ["participant_id", "group", "episode", "stimulus_id"]:
        data[c] = pd.Categorical(data[c])
    return data, exp, int(((data.raw_rating < 0) | (data.raw_rating > 100)).sum())
def flat(da):
    return da.stack(sample=("chain", "draw")).transpose("sample", ...)
def arr(idata, var):
    return flat(idata.posterior[var]).values
def hdi(x):
    lo, hi = az.hdi(np.asarray(x), hdi_prob=0.94)
    return float(lo), float(hi)
def diag(idata):
    s = az.summary(idata, hdi_prob=0.94)
    stats = idata.sample_stats
    return {
        "divergences": int(stats["diverging"].sum().item()),
        "max_rhat": float(s.r_hat.max()),
        "min_bulk_ess": float(s.ess_bulk.min()),
        "min_tail_ess": float(s.ess_tail.min()),
        "max_treedepth_warnings": int(stats["maxdepth_reached"].sum().item()) if "maxdepth_reached" in stats else 0,
    }
def fit_model(data, seed, config):
    m = bmb.Model(FORMULA, data, family="gaussian")
    t = time.perf_counter()
    idata = m.fit(progressbar=False, random_seed=seed, **config)
    return idata, time.perf_counter() - t
def fixed_recovery(idata, s):
    p = idata.posterior
    draws = {"z_RMS": arr(idata, "z_RMS"), "z_CF": arr(idata, "z_CF"), "z_SW": arr(idata, "z_SW")}
    ep = flat(p["episode"])
    for level in ep.coords["episode_dim"].values:
        draws[f"episode[{level}]"] = ep.sel(episode_dim=level).values
    gr = flat(p["group"])
    for level in gr.coords["group_dim"].values:
        draws[f"group[{level}]"] = gr.sel(group_dim=level).values
    rows = []
    for term, fn in TRUE_TERMS.items():
        vals = np.asarray(draws[term])
        lo, hi = hdi(vals)
        true = fn(s)
        mean = float(vals.mean())
        rows.append({"parameter": term, "true_value": true, "posterior_mean": mean, "posterior_sd": float(vals.std(ddof=1)), "hdi_3%": lo, "hdi_97%": hi, "bias": mean - true, "absolute_error": abs(mean - true), "squared_error": (mean - true)**2, "covered": lo <= true <= hi, "sign_recovered": np.sign(mean) == np.sign(true), "hdi_width": hi - lo})
    return pd.DataFrame(rows)
def icc_recovery(idata, s):
    pv, sv, rv = arr(idata, "1|participant_id_sigma")**2, arr(idata, "1|stimulus_id_sigma")**2, arr(idata, "sigma")**2
    total = pv + sv + rv
    ttotal = s["participant_sd"]**2 + s["stimulus_sd"]**2 + s["residual_sd"]**2
    draws = {"participant_ICC": pv/total, "stimulus_ICC": sv/total, "residual_share": rv/total}
    truth = {"participant_ICC": s["participant_sd"]**2/ttotal, "stimulus_ICC": s["stimulus_sd"]**2/ttotal, "residual_share": s["residual_sd"]**2/ttotal}
    rows = []
    for k, vals in draws.items():
        lo, hi = hdi(vals); mean = float(vals.mean()); true = truth[k]
        rows.append({"parameter": k, "true_value": true, "posterior_mean": mean, "posterior_sd": float(vals.std(ddof=1)), "hdi_3%": lo, "hdi_97%": hi, "bias": mean-true, "absolute_error": abs(mean-true), "squared_error": (mean-true)**2, "covered": lo <= true <= hi, "hdi_width": hi-lo})
    return pd.DataFrame(rows)
def expected_draws(idata, exp):
    p = idata.posterior
    intercept, zr, zc, zw = arr(idata, "Intercept"), arr(idata, "z_RMS"), arr(idata, "z_CF"), arr(idata, "z_SW")
    ep = flat(p["episode"]); epd = {str(x): ep.sel(episode_dim=x).values for x in ep.coords["episode_dim"].values}
    gr = flat(p["group"]); grd = {str(x): gr.sel(group_dim=x).values for x in gr.coords["group_dim"].values}
    st = flat(p["1|stimulus_id"]); std = {str(x): st.sel(stimulus_id__factor_dim=x).values for x in st.coords["stimulus_id__factor_dim"].values}
    rows, mat = [], []
    for _, r in exp.iterrows():
        vals = intercept.copy()
        vals = vals + (epd[r.episode] if r.episode in epd else 0) + (grd[r.group] if r.group in grd else 0)
        vals = vals + zr*r.z_RMS + zc*r.z_CF + zw*r.z_SW + std[r.stimulus_id]
        lo, hi = hdi(vals); mean = float(vals.mean())
        rows.append({**r[["song_id", "song_title", "group", "episode", "stimulus_id", "mix_id", "original_mix_name", "true_expected_rating"]].to_dict(), "posterior_mean_expected_rating": mean, "posterior_sd_expected_rating": float(vals.std(ddof=1)), "hdi_3%": lo, "hdi_97%": hi, "absolute_error": abs(mean-r.true_expected_rating), "squared_error": (mean-r.true_expected_rating)**2, "covered": lo <= r.true_expected_rating <= hi, "hdi_width": hi-lo})
        mat.append(vals)
    return pd.DataFrame(rows), np.vstack(mat)
def winners(cells, mat):
    rows = []
    cells = cells.reset_index(drop=True)
    for (song, ep), idx0 in cells.groupby(["song_id", "episode"]).groups.items():
        idx = list(idx0)
        truth = cells.loc[idx, "true_expected_rating"].to_numpy()
        true_pos = set(np.where(np.isclose(truth, truth.max()))[0].tolist())
        vals = mat[idx, :]
        ties = np.isclose(vals, vals.max(axis=0), atol=1e-12)
        probs = (ties / ties.sum(axis=0)).mean(axis=1)
        pred = int(np.where(np.isclose(probs, probs.max(), atol=1e-12))[0][0])
        rows.append({"song_id": song, "episode": ep, "n_candidates": len(idx), "winner_accuracy": pred in true_pos, "posterior_probability_true_winner": float(probs[list(true_pos)].sum()), "true_winner_probability_gt_0_50": float(probs[list(true_pos)].sum()) > 0.5, "draw_level_tie_rate": float((ties.sum(axis=0) > 1).mean()), "final_probability_tie": bool(np.isclose(probs, probs.max(), atol=1e-12).sum() > 1), "probability_sum": float(probs.sum())})
    return pd.DataFrame(rows)
def summarise(idata, scenario, exp, clipped):
    d = diag(idata)
    fr = fixed_recovery(idata, scenario)
    ir = icc_recovery(idata, scenario)
    ce, mat = expected_draws(idata, exp)
    wr = winners(ce, mat)
    feat = fr[fr.parameter.isin(FEATURE_TERMS)]
    picc = ir[ir.parameter.eq("participant_ICC")].iloc[0]
    sicc = ir[ir.parameter.eq("stimulus_ICC")].iloc[0]
    metrics = {
        "clipped_rating_count": clipped,
        "operational_success": d["divergences"] == 0 and d["max_rhat"] <= 1.05 and d["min_bulk_ess"] >= 50 and d["min_tail_ess"] >= 50,
        "strong_convergence": d["divergences"] == 0 and d["max_rhat"] <= 1.01 and d["min_bulk_ess"] >= 100 and d["min_tail_ess"] >= 100,
        "feature_coefficient_rmse": float(np.sqrt(feat.squared_error.mean())),
        "feature_coefficient_mae": float(feat.absolute_error.mean()),
        "feature_hdi_coverage": float(feat.covered.mean()),
        "feature_sign_recovery": float(feat.sign_recovered.mean()),
        "mean_feature_hdi_width": float(feat.hdi_width.mean()),
        "mean_fixed_posterior_sd": float(fr.posterior_sd.mean()),
        "participant_icc_bias": float(picc.bias),
        "participant_icc_squared_error": float(picc.squared_error),
        "participant_icc_covered": bool(picc.covered),
        "participant_icc_hdi_width": float(picc.hdi_width),
        "stimulus_icc_bias": float(sicc.bias),
        "stimulus_icc_squared_error": float(sicc.squared_error),
        "stimulus_icc_covered": bool(sicc.covered),
        "stimulus_icc_hdi_width": float(sicc.hdi_width),
        "cell_mean_rmse": float(np.sqrt(ce.squared_error.mean())),
        "cell_mean_hdi_coverage": float(ce.covered.mean()),
        "mean_cell_hdi_width": float(ce.hdi_width.mean()),
        "winner_accuracy": float(wr.winner_accuracy.mean()),
        "posterior_probability_true_winner": float(wr.posterior_probability_true_winner.mean()),
        "true_winner_probability_gt_0_50": float(wr.true_winner_probability_gt_0_50.mean()),
        "draw_level_tie_rate": float(wr.draw_level_tie_rate.mean()),
        "final_probability_tie_rate": float(wr.final_probability_tie.mean()),
    }
    return d, metrics, fr, ir, ce, wr

compat_payload = {"design_version": DESIGN_VERSION, "formula": FORMULA, "sample_sizes": SAMPLE_SIZES, "replications": N_REPLICATIONS, "scenarios": SCENARIOS, "feature_sha256": hashlib.sha256(FEATURE_TABLE_PATH.read_bytes()).hexdigest(), "sampling": SIM_CONFIG}
compat_hash = hashlib.sha256(json.dumps(compat_payload, sort_keys=True, default=str).encode()).hexdigest()
checkpoint = OUTPUT_DIR / "simulation_run_results_checkpoint_final_design.csv"
manifest = OUTPUT_DIR / "simulation_checkpoint_manifest.json"
if checkpoint.exists():
    try:
        old_manifest = json.loads(manifest.read_text(encoding="utf-8"))
        completed = pd.read_csv(checkpoint) if old_manifest.get("compatibility_hash") == compat_hash else pd.DataFrame()
    except Exception:
        completed = pd.DataFrame()
else:
    completed = pd.DataFrame()
manifest.write_text(json.dumps({**compat_payload, "compatibility_hash": compat_hash, "replication_fallback_reason": REPLICATION_FALLBACK_REASON}, indent=2), encoding="utf-8")
LONG_CHECKPOINT_FILES = ["fixed_effect_recovery_long.csv", "icc_recovery_long.csv", "cell_mean_recovery_long.csv", "winner_recovery_long.csv"]
for stale in OUTPUT_DIR.glob("*.csv"):
    if stale.name not in ["simulation_run_results_checkpoint_final_design.csv", *LONG_CHECKPOINT_FILES]:
        stale.unlink()
for stale in OUTPUT_DIR.glob("*.json"):
    if stale.name != "simulation_checkpoint_manifest.json":
        stale.unlink()
for stale in OUTPUT_DIR.glob("*.nc"):
    stale.unlink()
for stale in FIGURE_DIR.glob("*.png"):
    stale.unlink()
keys = set(zip(completed.scenario, completed.n_participants, completed.replication)) if not completed.empty else set()
def load_long_checkpoint(filename):
    path = OUTPUT_DIR / filename
    if not completed.empty and path.exists():
        return pd.read_csv(path).to_dict("records")
    return []
runs = completed.to_dict("records") if not completed.empty else []
failed_rows = []
fixed_rows = load_long_checkpoint("fixed_effect_recovery_long.csv")
icc_rows = load_long_checkpoint("icc_recovery_long.csv")
cell_rows = load_long_checkpoint("cell_mean_recovery_long.csv")
winner_rows = load_long_checkpoint("winner_recovery_long.csv")
planned = [(sc, n, r) for sc in SCENARIOS for n in SAMPLE_SIZES for r in range(1, N_REPLICATIONS+1)]
print(f"Planned fits: {len(planned)}; completed checkpoint fits: {len(keys)}")
for sc, n, rep in planned:
    if (sc, n, rep) in keys:
        continue
    print(f"Fitting {sc}, N={n}, replication={rep}")
    try:
        data, exp, clipped = generate(n, sc, rep)
        idata, runtime = fit_model(data, SEED + 1000000 + 10000*list(SCENARIOS).index(sc) + 100*n + rep, SIM_CONFIG)
        d, metrics, fr, ir, ce, wr = summarise(idata, SCENARIOS[sc], exp, clipped)
        base = {"design_version": DESIGN_VERSION, "scenario": sc, "n_participants": n, "replication": rep, "rows": len(data), "ratings_per_participant": 30, "runtime_seconds": runtime, "sampler": SIM_CONFIG["inference_method"], **{k: SIM_CONFIG[k] for k in ["draws", "tune", "chains", "cores", "target_accept"]}, **d, **metrics}
        runs.append(base)
        pd.DataFrame(runs).to_csv(checkpoint, index=False)
        fixed_rows += [{**x, "scenario": sc, "n_participants": n, "replication": rep} for x in fr.to_dict("records")]
        icc_rows += [{**x, "scenario": sc, "n_participants": n, "replication": rep} for x in ir.to_dict("records")]
        cell_rows += [{**x, "scenario": sc, "n_participants": n, "replication": rep} for x in ce.to_dict("records")]
        winner_rows += [{**x, "scenario": sc, "n_participants": n, "replication": rep} for x in wr.to_dict("records")]
        pd.DataFrame(fixed_rows).to_csv(OUTPUT_DIR / "fixed_effect_recovery_long.csv", index=False)
        pd.DataFrame(icc_rows).to_csv(OUTPUT_DIR / "icc_recovery_long.csv", index=False)
        pd.DataFrame(cell_rows).to_csv(OUTPUT_DIR / "cell_mean_recovery_long.csv", index=False)
        pd.DataFrame(winner_rows).to_csv(OUTPUT_DIR / "winner_recovery_long.csv", index=False)
    except Exception as e:
        failed_rows.append({"design_version": DESIGN_VERSION, "scenario": sc, "n_participants": n, "replication": rep, "error": repr(e)})
        pd.DataFrame(failed_rows).to_csv(OUTPUT_DIR / "failed_fits.csv", index=False)
        print("FAILED", sc, n, rep, e)

simulation_run_results = pd.DataFrame(runs)
failed_fits = pd.DataFrame(failed_rows)
fixed_long = pd.DataFrame(fixed_rows)
icc_long = pd.DataFrame(icc_rows)
cell_long = pd.DataFrame(cell_rows)
winner_long = pd.DataFrame(winner_rows)
print("Completed fits:", len(simulation_run_results), "Failed:", len(failed_fits))
display(simulation_run_results.tail())

metrics = ["operational_success","strong_convergence","feature_coefficient_rmse","feature_coefficient_mae","feature_hdi_coverage","feature_sign_recovery","mean_fixed_posterior_sd","mean_feature_hdi_width","participant_icc_bias","participant_icc_squared_error","participant_icc_covered","participant_icc_hdi_width","stimulus_icc_bias","stimulus_icc_squared_error","stimulus_icc_covered","stimulus_icc_hdi_width","cell_mean_rmse","cell_mean_hdi_coverage","mean_cell_hdi_width","winner_accuracy","posterior_probability_true_winner","true_winner_probability_gt_0_50","draw_level_tie_rate","final_probability_tie_rate","runtime_seconds"]
agg = []
for (sc, n), g in simulation_run_results.groupby(["scenario", "n_participants"]):
    row = {"scenario": sc, "n_participants": n, "n_completed": len(g)}
    for m in metrics:
        v = pd.to_numeric(g[m], errors="coerce").dropna().astype(float)
        row[f"{m}_mean"] = float(v.mean()) if len(v) else np.nan
        row[f"{m}_sd"] = float(v.std(ddof=1)) if len(v) > 1 else np.nan
        row[f"{m}_se"] = float(v.std(ddof=1) / math.sqrt(len(v))) if len(v) > 1 else np.nan
        row[f"{m}_p10"] = float(v.quantile(.1)) if len(v) else np.nan
        row[f"{m}_p90"] = float(v.quantile(.9)) if len(v) else np.nan
    row["participant_icc_rmse"] = math.sqrt(row["participant_icc_squared_error_mean"])
    row["stimulus_icc_rmse"] = math.sqrt(row["stimulus_icc_squared_error_mean"])
    agg.append(row)
sample_size_summary = pd.DataFrame(agg).sort_values(["scenario", "n_participants"])
icc_summary = sample_size_summary[["scenario","n_participants","n_completed","participant_icc_bias_mean","participant_icc_rmse","participant_icc_covered_mean","participant_icc_hdi_width_mean","stimulus_icc_bias_mean","stimulus_icc_rmse","stimulus_icc_covered_mean","stimulus_icc_hdi_width_mean"]].copy()
runtime_summary = simulation_run_results.groupby(["scenario","n_participants"]).agg(completed_fits=("replication","count"), median_runtime_seconds=("runtime_seconds","median"), p90_runtime_seconds=("runtime_seconds", lambda x: x.quantile(.9)), max_runtime_seconds=("runtime_seconds","max"), operational_success_rate=("operational_success","mean"), strong_convergence_rate=("strong_convergence","mean")).reset_index()
focus = sample_size_summary[sample_size_summary.n_participants.isin([40,50,60])][["scenario","n_participants","n_completed","operational_success_mean","strong_convergence_mean","feature_coefficient_rmse_mean","participant_icc_rmse","stimulus_icc_rmse","cell_mean_rmse_mean","cell_mean_hdi_coverage_mean","winner_accuracy_mean","posterior_probability_true_winner_mean","mean_cell_hdi_width_mean","runtime_seconds_mean"]].copy()
changes = []
for sc, g in sample_size_summary.groupby("scenario"):
    g = g.sort_values("n_participants")
    prev = None
    for _, r in g.iterrows():
        if prev is not None:
            changes.append({"scenario": sc, "transition": f"{int(prev.n_participants)}->{int(r.n_participants)}", "from_n": int(prev.n_participants), "to_n": int(r.n_participants), "feature_coefficient_rmse_change": r.feature_coefficient_rmse_mean - prev.feature_coefficient_rmse_mean, "participant_icc_rmse_change": r.participant_icc_rmse - prev.participant_icc_rmse, "cell_mean_rmse_change": r.cell_mean_rmse_mean - prev.cell_mean_rmse_mean, "winner_accuracy_change": r.winner_accuracy_mean - prev.winner_accuracy_mean, "mean_cell_hdi_width_change": r.mean_cell_hdi_width_mean - prev.mean_cell_hdi_width_mean}
            )
        prev = r
diminishing_returns = pd.DataFrame(changes)
tmp = diminishing_returns.copy()
tmp["mean_improvement"] = (-tmp.feature_coefficient_rmse_change - tmp.participant_icc_rmse_change - tmp.cell_mean_rmse_change - tmp.mean_cell_hdi_width_change) / 4
diminishing_return_point = tmp.groupby("transition").mean_improvement.mean().sort_values().index[0]
display(sample_size_summary[["scenario","n_participants","n_completed","operational_success_mean","feature_coefficient_rmse_mean","participant_icc_rmse","stimulus_icc_rmse","cell_mean_rmse_mean","winner_accuracy_mean","mean_cell_hdi_width_mean"]])
display(focus)
display(diminishing_returns)

scores = []
for n, g in sample_size_summary.groupby("n_participants"):
    scores.append({"n_participants": int(n), "mean_operational_success": g.operational_success_mean.mean(), "mean_feature_rmse": g.feature_coefficient_rmse_mean.mean(), "mean_participant_icc_rmse": g.participant_icc_rmse.mean(), "mean_stimulus_icc_rmse": g.stimulus_icc_rmse.mean(), "mean_cell_rmse": g.cell_mean_rmse_mean.mean(), "mean_winner_accuracy": g.winner_accuracy_mean.mean(), "mean_true_winner_probability": g.posterior_probability_true_winner_mean.mean(), "mean_cell_hdi_width": g.mean_cell_hdi_width_mean.mean(), "mean_runtime_seconds": g.runtime_seconds_mean.mean()})
candidate_scores = pd.DataFrame(scores).sort_values("n_participants")
n20_width = candidate_scores.loc[candidate_scores.n_participants.eq(20), "mean_cell_hdi_width"].iloc[0]
candidate_scores["interval_width_reduction_vs_20"] = 1 - candidate_scores.mean_cell_hdi_width / n20_width
candidate_scores["adequate"] = (candidate_scores.mean_operational_success >= .90) & (candidate_scores.mean_cell_rmse <= 4.0) & (candidate_scores.mean_winner_accuracy >= .70) & (candidate_scores.interval_width_reduction_vs_20 >= .20)
candidate_final_n = int(candidate_scores.loc[candidate_scores.adequate, "n_participants"].iloc[0]) if candidate_scores.adequate.any() else int(candidate_scores[candidate_scores.n_participants.isin([40,50,60])].sort_values(["mean_cell_rmse","n_participants"]).iloc[0].n_participants)
display(candidate_scores)
print("Evidence-based candidate final N:", candidate_final_n)

def plot_metric(metric, ylabel, filename):
    fig, ax = plt.subplots(figsize=(8, 5))
    for sc, g in sample_size_summary.groupby("scenario"):
        g = g.sort_values("n_participants")
        ax.plot(g.n_participants, g[metric], marker="o", label=sc)
    ax.set_xlabel("Participants"); ax.set_ylabel(ylabel); ax.set_title(ylabel + " by Sample Size"); ax.legend(frameon=True)
    fig.tight_layout()
    p = FIGURE_DIR / filename
    fig.savefig(p, dpi=180)
    plt.show()
    return p
figures = [
    plot_metric("feature_coefficient_rmse_mean", "Feature-Coefficient RMSE", "participant_count_vs_feature_coefficient_rmse.png"),
    plot_metric("participant_icc_rmse", "Participant-ICC RMSE", "participant_count_vs_participant_icc_rmse.png"),
    plot_metric("cell_mean_rmse_mean", "Stimulus x Episode Mean RMSE", "participant_count_vs_cell_mean_rmse.png"),
    plot_metric("winner_accuracy_mean", "Preferred-Mix Winner Accuracy", "participant_count_vs_winner_accuracy.png"),
    plot_metric("mean_cell_hdi_width_mean", "Mean Expected-Rating HDI Width", "participant_count_vs_interval_width.png"),
]
fig, ax = plt.subplots(figsize=(8, 5))
for sc, g in diminishing_returns.groupby("scenario"):
    ax.plot(g.transition, -g.cell_mean_rmse_change, marker="o", label=sc)
ax.axhline(0, color="black", linewidth=.8); ax.set_xlabel("Transition"); ax.set_ylabel("Cell RMSE improvement"); ax.set_title("Diminishing Returns"); ax.legend(frameon=True)
fig.tight_layout(); p = FIGURE_DIR / "diminishing_returns_cell_rmse.png"; fig.savefig(p, dpi=180); plt.show(); figures.append(p)

confirmation_data, confirmation_exp, confirmation_clipped = generate(candidate_final_n, "moderate_expected", 999)
confirmation_idata, confirmation_runtime = fit_model(confirmation_data, SEED + 999999, CONFIRM_CONFIG)
cd, cm, cfr, cir, cce, cwr = summarise(confirmation_idata, SCENARIOS["moderate_expected"], confirmation_exp, confirmation_clipped)
ppc_outside = np.nan
try:
    model_for_ppc = bmb.Model(FORMULA, confirmation_data, family="gaussian")
    model_for_ppc.predict(confirmation_idata, kind="pps", inplace=True)
    vals = confirmation_idata.posterior_predictive["rating"].values
    ppc_outside = float(((vals < 0) | (vals > 100)).mean())
except Exception as e:
    print("PPC generation failed:", e)
confirmation_summary = pd.DataFrame([{**{"design_version": DESIGN_VERSION, "scenario": "moderate_expected", "n_participants": candidate_final_n, "rows": len(confirmation_data), "runtime_seconds": confirmation_runtime, "sampler": CONFIRM_CONFIG["inference_method"], **CONFIRM_CONFIG}, **cd, **cm, "ppc_outside_0_100": ppc_outside}])
display(confirmation_summary); display(cfr); display(cir); display(cwr)

scenario_table.to_csv(OUTPUT_DIR / "simulation_scenarios.csv", index=False)
checks.to_csv(OUTPUT_DIR / "feature_table_validation.csv", index=False)
simulation_run_results.to_csv(OUTPUT_DIR / "simulation_run_results.csv", index=False)
sample_size_summary.to_csv(OUTPUT_DIR / "sample_size_summary.csv", index=False)
icc_summary.to_csv(OUTPUT_DIR / "icc_recovery_summary.csv", index=False)
diminishing_returns.to_csv(OUTPUT_DIR / "diminishing_returns.csv", index=False)
focus.to_csv(OUTPUT_DIR / "sample_size_40_50_60_comparison.csv", index=False)
failed_fits.to_csv(OUTPUT_DIR / "failed_fits.csv", index=False)
runtime_summary.to_csv(OUTPUT_DIR / "runtime_summary.csv", index=False)
fixed_long.to_csv(OUTPUT_DIR / "fixed_effect_recovery_long.csv", index=False)
icc_long.to_csv(OUTPUT_DIR / "icc_recovery_long.csv", index=False)
cell_long.to_csv(OUTPUT_DIR / "cell_mean_recovery_long.csv", index=False)
winner_long.to_csv(OUTPUT_DIR / "winner_recovery_long.csv", index=False)
confirmation_summary.to_csv(OUTPUT_DIR / "confirmation_fit_summary.csv", index=False)
cfr.to_csv(OUTPUT_DIR / "confirmation_fit_fixed_effect_recovery.csv", index=False)
cir.to_csv(OUTPUT_DIR / "confirmation_fit_icc_recovery.csv", index=False)
cwr.to_csv(OUTPUT_DIR / "confirmation_fit_winner_recovery.csv", index=False)
recommendation = {
    "design_version": DESIGN_VERSION, "final_model_formula": FORMULA, "sample_sizes_tested": SAMPLE_SIZES,
    "scenarios_tested": list(SCENARIOS.keys()), "requested_replications_per_condition": REQUESTED_REPLICATIONS,
    "actual_replications_per_condition": N_REPLICATIONS, "replication_fallback_reason": REPLICATION_FALLBACK_REASON,
    "candidate_final_n": candidate_final_n, "diminishing_return_point": diminishing_return_point,
    "adequacy_criteria": ADEQUACY_CRITERIA, "candidate_scores": candidate_scores.to_dict("records"),
    "confirmation_fit": confirmation_summary.to_dict("records"),
    "unresolved_concerns": [
        "Only 10 replications per condition were executed, so Monte Carlo uncertainty remains visible.",
        "Group is structurally confounded with assigned song set.",
        "Scenarios are hypothetical and real participant data will require fresh diagnostics.",
    ],
}
(OUTPUT_DIR / "sample_size_recommendation.json").write_text(json.dumps(recommendation, indent=2), encoding="utf-8")
(OUTPUT_DIR / "report_ready_sample_size_justification.txt").write_text(
    f"Phase 3C final-design simulation used {FORMULA}. Candidate final N: {candidate_final_n}. "
    f"The requested 20 replications were reduced to {N_REPLICATIONS}: {REPLICATION_FALLBACK_REASON} "
    "The recommendation is based on convergence, ICC recovery, feature-effect recovery, expected-rating recovery, preferred-mix recovery, uncertainty trends, and diminishing returns.",
    encoding="utf-8",
)
exports = sorted([str(p.relative_to(REPO_ROOT)) for p in OUTPUT_DIR.glob("*.csv")]) + sorted([str(p.relative_to(REPO_ROOT)) for p in OUTPUT_DIR.glob("*.json")]) + sorted([str(p.relative_to(REPO_ROOT)) for p in FIGURE_DIR.glob("*.png")])
display(pd.DataFrame({"exported_file": exports}))

phase3c_summary = {
    "notebook": "statistical-baseline/notebooks/_05_sample_size_simulation.ipynb",
    "final_model_formula": FORMULA,
    "sample_sizes_tested": SAMPLE_SIZES,
    "scenarios_tested": list(SCENARIOS.keys()),
    "requested_replications_per_condition": REQUESTED_REPLICATIONS,
    "actual_replications_per_condition": N_REPLICATIONS,
    "total_fits_attempted": len(planned),
    "total_fits_completed": int(len(simulation_run_results)),
    "total_fits_failed": int(len(failed_fits)),
    "sampling_configuration": SIM_CONFIG,
    "runtime_distribution_seconds": {"median": float(simulation_run_results.runtime_seconds.median()), "p90": float(simulation_run_results.runtime_seconds.quantile(.9)), "max": float(simulation_run_results.runtime_seconds.max())},
    "operational_success_rate": float(simulation_run_results.operational_success.mean()),
    "strong_convergence_rate": float(simulation_run_results.strong_convergence.mean()),
    "candidate_final_n": int(candidate_final_n),
    "diminishing_return_point": diminishing_return_point,
    "confirmation_fit_diagnostics": confirmation_summary.to_dict("records"),
    "exported_files": exports,
    "unresolved_concerns": recommendation["unresolved_concerns"],
}
phase3c_summary
```

This notebook computes the evidence needed for Phase 3D but does not label Phase 3D complete. Adequacy is judged using convergence, participant/stimulus ICC recovery, acoustic coefficient recovery, expected-rating recovery, preferred-mix recovery, uncertainty trends, runtime, and diminishing returns. Group effects are reported cautiously because group is structurally linked to song allocation.
